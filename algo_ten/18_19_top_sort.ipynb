```python
# 2115. Find All Possible Recipes from Given Supplies
```

```python
class Solution:
    def findAllRecipes(self, recipes: list[str], ingredients: list[list[str]], supplies: list[str]) -> list[str]:
        in_degree = {}
        g = defaultdict(list)
        for recipe, components in zip(recipes, ingredients):
            in_degree[recipe] = len(components)
            for component in components:
                g[component].append(recipe)

        q = deque(supplies)
        result = []

        while q:
            supply = q.popleft()
            for recipe in g[supply]:
                in_degree[recipe] -= 1
                if in_degree[recipe] == 0:
                    q.append(recipe)
                    result.append(recipe)

        return result
```

```python
# 207. Course Schedule
```

```python
class Solution:
    def canFinish(self, numCourses: int, prerequisites: list[list[int]]) -> bool:
        # a <- b
        in_degree = defaultdict(int)
        g = defaultdict(list)
        for a, b in prerequisites:
            g[b].append(a)
            in_degree[a] += 1

        queue = deque()
        for i in range(numCourses):
            if in_degree[i] == 0:
                queue.append(i)
        
        n = 0

        while queue:
            node = queue.popleft() 
            n += 1

            for ngh in g[node]:
                in_degree[ngh] -= 1
                if in_degree[ngh]  == 0:
                    queue.append(ngh)

        return n == numCourses
```

```python
# 210. Course Schedule II
```

```python
class Solution:
    def findOrder(self, numCourses: int, prerequisites: list[list[int]]) -> list[int]:
        # a <- b
        g = defaultdict(list)
        in_degree = defaultdict(int)

        for a, b in prerequisites:
            g[b].append(a)
            in_degree[a] += 1
        
        queue = deque()
        for i in range(numCourses):
            if in_degree[i] == 0:
                queue.append(i)
        
        order = []

        while queue:
            node = queue.popleft()
            order.append(node)

            for ngh in g[node]:
                in_degree[ngh] -= 1
                if in_degree[ngh] == 0:
                    queue.append(ngh)
        if len(order) == numCourses:
            return order
        return []
```

```python
# 1136 parallel courses
```

```python
from collections import defaultdict, deque


class Solution:
    def minimunSemesters(self, n: int, relations: list[list[int]]) -> int:
        g = defaultdict(list)
        in_degree = defaultdict(int)

        for a, b in relations:
            g[a].append(b)
            in_degree[b] += 1

        queue = deque()

        for i in range(1, n + 1):
            if in_degree[i] == 0:
                queue.append(i)

        semesters = 0
        result = 0

        while queue:
            level = len(queue)
            for _ in range(level):
                node = queue.popleft()
                n -= 1
                for ngh in g[node]:
                    in_degree[ngh] -= 1
                    if in_degree[ngh] == 0:
                        queue.append(ngh)

            semesters += 1

        if n == 0:
            return semesters

        return -1
```
