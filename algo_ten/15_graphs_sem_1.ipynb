```python
# 200. Number of Islands
```

```python
class Solution:
    def numIslands(self, grid: list[list[str]]) -> int:
        n = len(grid)
        m = len(grid[0])

        number = 0

        def isNotvalid(r, c):
            return r < 0 or c < 0 or r >= n or c >= m or grid[r][c] != "1"

        def isValid(r, c):
            return 0 <= r and 0 <= c and r < n or c < m and grid[r][c] == "1"

        for i in range(n):
            for j in range(m):
                if grid[i][j] == "1":
                    number += 1
                    queue = deque([(i, j)])
                    while queue:
                        r, c = queue.popleft()
                        if not isNotvalid(r, c):
                            grid[r][c] = "0"
                            queue.append((r + 1, c))
                            queue.append((r - 1, c))
                            queue.append((r, c + 1))
                            queue.append((r, c - 1))

        return number
```

```python
# 1466. Reorder Routes to Make All Paths Lead to the City Zero
```

```python
class Solution:
    def minReorder(self, n: int, connections: list[list[int]]) -> int:
        og_directions = set()
        graph = defaultdict(list)
        for a, b in connections:
            graph[a].append(b)
            graph[b].append(a)
            og_directions.add((a, b))

        seen = {0}
        turns = 0
        stack = [0]
        while stack:
            v = stack.pop()
            for n in graph[v]:
                if n not in seen:
                    if (v, n) in og_directions:
                        turns += 1 
                    stack.append(n)
                    seen.add(n)

        return turns
```

```python
# 695. Max Area of Island
```

```python
class Solution:
    def maxAreaOfIsland(self, grid: list[list[int]]) -> int:
        n = len(grid)
        m = len(grid[0])

        def is_not_valid(r, c):
            return r < 0 or c < 0 or r >= n or c >= m or grid[r][c] != 1

        max_area = 0

        for i in range(n):
            for j in range(m):
                if grid[i][j] == 1:
                    cur_area = 0
                    queue = deque([(i, j)])
                    grid[i][j] = 0 
                    
                    while queue:
                        r, c = queue.popleft() 
                        cur_area += 1
                        
                        for dx, dy in [(0, 1), (1, 0), (-1, 0), (0, -1)]:
                            nr, nc = r + dx, c + dy 
                            if not is_not_valid(nr, nc):
                                queue.append((nr, nc))
                                grid[nr][nc] = 0
                                
                    max_area = max(max_area, cur_area)

        return max_area
```

```python
# 2368. Reachable Nodes With Restrictions
```

```python
class Solution:
    def reachableNodes(self, n: int, edges: list[list[int]], restricted: list[int]) -> int:
        graph = defaultdict(list)
        for a, b in edges:
            graph[a].append(b)
            graph[b].append(a)

        seen = set(restricted)
        seen.add(0)

        stack = [0]
        result = 0
        while stack:
            v = stack.pop()
            result += 1
            for n in graph[v]:
                if n not in seen:
                    stack.append(n)
                    seen.add(n)
        return result
```

```python
# 542. 01 Matrix
```

```python
class Solution:
    def updateMatrix(self, mat: list[list[int]]) -> list[list[int]]:
        if not mat:
            return mat

        n = len(mat)
        m = len(mat[0])

        seen = set()

        def is_valid(r, c):
            return 0 <= r and 0 <= c and r < n and c < m and (r, c) not in seen

        queue = deque([])

        for i in range(n):
            for j in range(m):
                if mat[i][j] == 0:
                    queue.append((i, j, 1))
                    seen.add((i, j))

        while queue:
            r, c, d = queue.popleft()
            for dx, dy in [(0, 1), (1, 0), (-1, 0), (0, -1)]:
                if is_valid(r+dx, c+dy):
                    mat[r+dx][c+dy] = d
                    queue.append((r + dx, r + dy, d + 1))
                    seen.add((r+dx, c + dy))

        return mat
```

```python
# 994. Rotting Oranges
```

```python
class Solution:
    def orangesRotting(self, grid: list[list[int]]) -> int:
        n = len(grid)
        m = len(grid[0])
        queue = deque([])
        fresh_counter = 0
        for i in range(n):
            for j in range(m):
                if grid[i][j] == 2:
                    queue.append((i, j))
                elif grid[i][j] == 1:
                    fresh_counter += 1

        if fresh_counter == 0:
            return 0

        ticks = 0
        directions = [(0, 1), (1, 0), (-1, 0), (0, -1)]
        def is_valid(r, c):
            return r >= 0 and c >= 0 and r < n and c < m and grid[r][c] == 1
            
        while queue and fresh_counter > 0:
            ticks += 1
            for _ in range(len(queue)):
                r, c = queue.popleft()
                for dx, dy in directions:
                    nr, nc = r + dx, c + dy

                    if is_valid(nr, nc):
                        grid[nr][nc] = 2
                        fresh_counter -= 1
                        queue.append((nr, nc)) 

        return ticks if fresh_counter == 0 else -1
```
