```python
from collections import defaultdict
import heapq
```

```python
# 743. Network Delay Time
```

```python
class Solution:
    def networkDelayTime(self, times: list[list[int]], n: int, k: int) -> int:
        # 1. Строим граф
        graph = defaultdict(list)
        for a, b, w in times:
            graph[a].append((b, w))

        # 2. Инициализируем кучу и словарь расстояний
        min_heap = [(0, k)]
        distances = {}
        distances[k] = 0

        while min_heap:
            dst, node = heapq.heappop(min_heap)
            
            if dst > distances.get(node, float("-inf")):
                continue

            for ngh, dst_to_ngh in graph[node]:
                new_dst = dst + dst_to_ngh
                if new_dst < distances.get(ngh, float("inf")):
                    distances[ngh] = new_dst
                    heapq.heappush(min_heap, (new_dst, ngh))
        if n != len(distances):
            return -1 

        return distances[max(distances, key = lambda x: distances[x])]
```

```python
# 1514. Path with Maximum Probability
```

```python
class Solution:
    def maxProbability(self, n: int, edges: list[list[int]], succProb: list[float], start_node: int, end_node: int) -> float:
        graph = defaultdict(list)
        for i in range(len(edges)):
            a = edges[i][0]
            b = edges[i][1]
            p = succProb[i]
            graph[a].append((b, p))
            graph[b].append((b, p))

        max_heap = [(-1.0, start_node)]
        probs = {}
        probs[start_node] = 1.0

        while max_heap:
            p, node = heapq.heappop(max_heap)
            if node == end_node:
                return -p
            
            for ngh, p_to_ngh in graph[node]:
                new_p = -p * p_to_ngh
                if ngh not in probs or new_p > probs[ngh]:
                    probs[ngh] = new_p
                    heapq.heappush(max_heap, (-new_p, ngh))
 
        return 0
```
