```python
# 26. Remove Duplicates from Sorted Array
```

```python
def canAttendMeetings(intervals):
    intervals.sort()

    for i in range(1, len(intervals)):
        prev = intervals[i - 1]
        curr = intervals[i]
        if curr[0] < prev[1]:
            return False
           
    return True
```

```python
def test_canAttendMeetings():
    assert canAttendMeetings([[0, 30], [5, 10], [15, 20]]) == False
    assert canAttendMeetings([[7, 10], [2, 4]]) == True
    assert canAttendMeetings([[1, 5], [5, 6]]) == True
    assert canAttendMeetings([[1, 3], [2, 4], [3, 5]]) == False
    assert canAttendMeetings([[1, 2], [2, 3], [3, 4]]) == True

test_canAttendMeetings()
```

```python
def merge(intervals):
    if not intervals:
        return []

    intervals.sort(key=lambda x: x[0])
    merged = [intervals[0]]

    for current in intervals[1:]:
        last_merged = merged[-1]
        if current[0] <= last_merged[1]:  # Overlapping intervals
            last_merged[1] = max(last_merged[1], current[1])  # Merge
        else:
            merged.append(current)

    return merged
```

```python
def test_merge():
    assert merge([[1, 3], [2, 6], [8, 10], [15, 18]]) == [[1, 6], [8, 10], [15, 18]]
    assert merge([[1, 4], [4, 5]]) == [[1, 5]]
    assert merge([[1, 4], [0, 2], [3, 5]]) == [[0, 5]]
    assert merge([[1, 4]]) == [[1, 4]]
    assert merge([]) == []

test_merge()
```
