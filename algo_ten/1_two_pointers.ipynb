344. Reverse String

```python
def revers_string(s: list[str]) -> str:
    l, r = 0, len(s) - 1
    while l < r:
        s[l], s[r] = s[r], s[l]
        l += 1
        r -= 1
    return s
```

```python
revers_string(["h", "e", "l", "o", "w"])
```

125. Valid Palindrome

```python
def isPalindrome(s: str) -> bool:
    l, r = 0, len(s) - 1
    while l < r:
        while l < r and not s[l].isalnum():
            l += 1
        while l < r and not s[r].isalnum():
            r -= 1
        if s[l].lower() != s[r].lower():
            return False
        l += 1
        r -= 1
    return True
```

```python
isPalindrome("A man, a plan, a canal: Panama")
```

167. Two Sum II - Input Array Is Sorted

```python
def find_sum(arr: list[int], s: int):
    arr.sort()
    l, r = 0, len(arr) - 1

    while l < r:
        cur_sum = l + r
        if cur_sum > s:
            r -= 1
        elif cur_sum < s:
            l += 1
        else:
            return arr[l], arr[r]

    return None
```

```python
find_sum([1, 2, 3, 4, 5, 6], 2)
```

```python
def two_sum(arr: list[int], target: int) -> list[int]:
    arr.sort()
    l, r = 0, len(arr) - 1

    while l < r:
        cur_sum = arr[l] + arr[r]
        if cur_sum == target:
            return [arr[l], arr[r]]
        if cur_sum < target:
            l += 1
        else:
            r -= 1

    return []
```

```python
two_sum([2, 7, 11, 15], 18)
```

15. 3Sum

```python
class Solution:
    def threeSum(self, nums: list[int]) -> list[list[int]]:
        nums = sorted(nums)
        l, r = 0, len(nums) - 1

        while l < r:
            cur_sum = nums[l] + nums[r]

            for i in range(len(nums)):
                if cur_sum == -nums[i] and nums[l] != nums[i] and nums[r] != nums[i]:
                    return [nums[l], nums[r], nums[i]]

            l += 1
            r -= 1

        return []

    def threeSum_v2(self, nums: list[int]) -> list[list[int]]:
        nums = sorted(nums)
        n = len(nums)
        res = set()

        for i in range(n):
            target = -nums[i]
            l, r = i + 1, n - 1
            while l < r:
                cur_sum = nums[l] + nums[r]

                if cur_sum == target:
                   res.add((nums[i], nums[l], nums[r])) 
                if cur_sum < target:
                    l += 1
                else:
                    r -= 1

        return res
```

977. Squares of a Sorted Array

```python
class Solution:
    def sortedSquares(self, nums: list[int]) -> list[int]:
        nums = sorted(nums)

        l, r = 0, len(nums) - 1
        res = [0] * len(nums)
        for i in range(len(nums) - 1, -1, -1):
            if abs(nums[l]) < abs(nums[r]):
                res[i] = nums[r]**2
                r -= 1
            else:
                res[i] = nums[l]**2
                l += 1

        return res
```

```python
nums = [-4,-1,0,3,10]
output = [0,1,9,16,100]

s = Solution()
result = s.sortedSquares(nums)
result == output
```

11. Container With Most Water

```python
class Solution:
    def maxArea(self, height: list[int]) -> int:
        n = len(height)
        l, r = 0, n - 1
        max_wather = 0
        while l < r:
            width = r - l
            cur_height = min(height[l], height[r])

            cur_wather = cur_height * width
            max_wather = max(max_wather, cur_wather)

            if height[l] < height[r]:
                l += 1
            else: 
                r -= 1

        return max_wather
```

```python
nums = [1,8,6,2,5,4,8,3,7]
output = 49

s = Solution()
result = s.maxArea(nums)
result == output
```

26. Remove Duplicates from Sorted Array

```python
class Solution:
    def removeDuplicates(self, nums: list[int]) -> int:
        k = 0
        for j in range(len(nums)):
            if nums[k] != nums[j]:
                k += 1
                nums[k] = nums[j]

        return k + 1

    def removeWithoutDuplicatesList(self, nums: list[int]) -> list[int]:
        k = 0

        for j in range(1, len(nums)):
            if nums[j] != nums[k]:
                k += 1
                nums[k] = nums[j]

        return nums[: k + 1]
```

```python
nums = [1,1,2, 2, 2, 3, 3, 4]
output = [1, 2]

s = Solution()
result = s.removeWithoutDuplicatesList(nums)
# result == output
result
```

283. Move Zeroes

```python
def moveZeroes(nums: list[int]) -> None:
    w = 0

    for r in range(len(nums)):
        if nums[r] != 0:
            nums[w], nums[r] = nums[r], nums[w]
            w += 1
```

844. Backspace String Compare

```python
class Solution:
    def backspaceCompare(self, s: str, t: str) -> bool:
        p1 = len(s) - 1
        p2 = len(t) - 1

        skip_s = 0
        skip_t = 0
        while p1 >= 0 and p2 <= 0:
            while p1 >= 0:
                if s[p1] == "#":
                    skip_s += 1
                    p1 -= 1
                elif skip_s > 0:
                    skip_s -= 1
                    p1 -= 1
                else:
                    break

            while p2 >= 0:
                if t[p2] == "#":
                    skip_t += 1
                    p2 -= 1
                elif skip_t > 0:
                    skip_t -= 1
                    p2 -= 1
                else:
                    break

            if p1 >= 0 and p2 >= 0:
                if s[p1] != t[p2]:
                    return False
            elif p1 >= 0 or p2 >= 0:
                return False

        return True
```

```python
s = "ab#c"

t = "ad#c"

res = Solution()
rez = res.backspaceCompare(s, t)

rez
```

88. Merge Sorted Array

```python
class Solution:
    def merge(self, nums1: list[int], m: int, nums2: list[int], n: int) -> None:
        p1 = 0
        p2 = 0
        result = []
        
        while p1 < m and p2 < n:
            if nums1[p1] < nums2[p2]:
                result.append(nums1[p1])
                p1 += 1
            else:
                result.append(nums2[p2])
                p2 += 1
                
        for k in range(p1, m):
            result.append(nums1[p1])
            p1 += 1   

        for k in range(p2, n):
            result.append(nums2[p2])
            p2 += 1   

        for i in range(len(result)):
            nums1[i] = result[i]
```

```python
nums1 = [1,2,3,0,0,0]
m = 3
nums2 = [2,5,6]
n = 3

sol = Solution()
res = sol.merge(nums1, m, nums2, n)
nums1
```
