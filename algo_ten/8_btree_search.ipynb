```python
# 700. Search in a Binary Search Tree
```

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


class Solution:
    def searchBST(self, root: TreeNode | None, val: int) -> TreeNode | None:
        if root and root.val == val:
            return root

        if not root:
            return

        if val < root.val:
            return self.searchBST(root.left, val)
        else:
            return self.searchBST(root.right, val)
```

```python
class Solution:
    def searchBST(self, root: TreeNode | None, val: int) -> TreeNode | None:
        stack = [root]

        while stack:
            node = stack.pop()
            if not node:
                continue
            if node.val == val:
                return node

            if val < node.val:
                stack.append(node.left)
            else:
                stack.append(node.right)

        return None
```

```python
# 701. Insert into a Binary Search Tree
```

```python
class Solution:
    def insertIntoBST(self, root: TreeNode | None, val: int) -> TreeNode | None:
        if not root:
            return TreeNode(val)

        if val < root.val:
            root.left = self.insertIntoBST(root.left, val)
        else:
            root.right = self.insertIntoBST(root.right, val)

        return root
```

```python
class Solution:
    def insertIntoBST(self, root: TreeNode | None, val: int) -> TreeNode | None:
        if not root:
            return TreeNode(val)

        stack = [root]

        while stack:
            node = stack.pop()
            if val < node.val:
                if not node.left:
                    node.left = TreeNode(val)
                    return root
                else:
                    stack.append(node.left)
            else:
                if not node.right:
                    node.right = TreeNode(val)
                    return root
                else:
                    stack.append(node.right)

        return root
```

```python
# 98. Validate Binary Search Tree
```

```python
class Solution:
    def isValidBST(self, root: TreeNode | None) -> bool:
        stack = [(root, float('-inf'), float('inf'))]

        while stack:
            node, min_r, max_r = stack.pop()
            if not node:
                continue
            if node.val <= min_r or node.val >= max_r:
                return False
            stack.append((node.left, min_r, node.val))
            stack.append((node.right, node.val, max_r))
        return True
```

```python
class Solution:
    def isValidBST(self, root: TreeNode | None) -> bool:
        
        # Вспомогательная функция, которая делает всю работу
        def validate(node, min_r, max_r):
            # 1. Если узла нет (дошли до пустого места), возвращаем True — нарушений нет
            if not node:
                return True
            
            # 2. Проверяем сам узел: если он вышел за рамки дозволенного, то это не BST
            if node.val <= min_r or node.val >= max_r:
                return False
            
            # 3. Делегируем задачу детям. 
            # Оба ребенка должны вернуть True, поэтому используем 'and'
            # Для левого ребенка: верхняя граница сужается до значения текущего узла
            # Для правого ребенка: нижняя граница поднимается до значения текущего узла
            return (validate(node.left, min_r, node.val) and 
                    validate(node.right, node.val, max_r))
        
        # Запускаем проверку с корня, давая ему бесконечные границы
        return validate(root, float('-inf'), float('inf'))
```

```python
# 110. Balanced Binary Tree
```

```python
class Solution:
    def isBalanced(self, root: TreeNode | None) -> bool:
        def height(root):
            if not root:
                return 0
            return 1 + max(height(root.left), height(root.right))
        
        if not root:
            return True
        
        left_h = height(root.left)
        right_h = height(root.right)

        if abs(left_h - right_h) > 1:
            return False
        return self.isBalanced(root.left) and self.isBalanced(root.right)
```

```python
class Solution:
    def isBalanced(self, root: TreeNode | None) -> bool:
        if not root:
            return True

        # Стек хранит кортежи: (узел, посещены_ли_дети)
        stack = [(root, False)]
        # Словарь для хранения высот узлов: {узел: высота}
        heights = {}

        while stack:
            node, visited = stack.pop()

            if not node:
                continue

            if not visited:
                # Помещаем родительский узел обратно, помечая, что теперь мы идем в его детей
                stack.append((node, True))
                # Добавляем детей (сначала правый, потом левый, чтобы левый извлекся первым)
                stack.append((node.right, False))
                stack.append((node.left, False))
            else:
                # Ветка visited == True срабатывает, когда мы поднимаемся обратно от детей к родителю
                left_height = heights.get(node.left, 0)
                right_height = heights.get(node.right, 0)

                # Если разница высот поддеревьев больше 1, дерево не сбалансировано
                if abs(left_height - right_height) > 1:
                    return False

                # Записываем высоту текущего узла для его родителя
                heights[node] = 1 + max(left_height, right_height)

        return True
```
